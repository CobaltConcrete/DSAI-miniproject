# Exploratory Data Analysis on BankChurners Dataset

This notebook provides a comprehensive EDA on the BankChurners dataset. The analysis includes data loading, summary statistics, missing value checks, distributions of numeric and categorical features, and a correlation matrix.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# For better plots inline in Jupyter Notebook
%matplotlib inline

plt.style.use('default')  # use default style to ensure no external styling
```

```python
# Load the dataset
data = pd.read_csv('BankChurners.csv')
print("Dataset loaded successfully!")
print(f"Shape of the dataset: {data.shape}")
```

## Data Overview

Let's take a look at the first few rows of the dataset and get a summary of its structure.

```python
# Display the first five rows
print("First 5 rows of the dataset:")
display(data.head())

# Display information about the dataset
print("\nDataset Info:")
data.info()

# Check for missing values
print("\nMissing values in each column:")
print(data.isnull().sum())
```

## Statistical Summary

Here we calculate the descriptive statistics for numerical columns.

```python
print("\nStatistical Summary for numeric columns:")
display(data.describe())
```

## Visualizations

We will now create some basic plots to visualize the data.

```python
# Identify numeric and categorical columns
numeric_cols = data.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = data.select_dtypes(include=['object']).columns.tolist()

print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
```

### Histograms for Numeric Features

We plot histograms to see the distributions of the numeric features.

```python
import math
import matplotlib.pyplot as plt

# numeric_cols is your list of numeric columns
n_cols = 3  # number of columns for subplots
n_rows = math.ceil(len(numeric_cols) / n_cols)  # automatically determine rows

plt.figure(figsize=(15, 5 * n_rows))  # height scales with number of rows

for i, col in enumerate(numeric_cols):
    plt.subplot(n_rows, n_cols, i + 1)
    plt.hist(data[col].dropna(), bins=20, edgecolor='black')
    plt.title(col)
    plt.tight_layout()

plt.show()
```

### Bar Plots for Categorical Features

We display the counts for each category in the categorical columns.

```python
plt.figure(figsize=(15, 12))
for i, col in enumerate(categorical_cols):
    plt.subplot(3, 3, i+1)  
    counts = data[col].value_counts()
    plt.bar(counts.index.astype(str), counts.values, edgecolor='black')
    plt.title(col)
    plt.xticks(rotation=45)
    plt.tight_layout()
plt.show()
```

### Correlation Matrix

Let's calculate and plot the correlation matrix for the numeric features.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Suppose 'data' is your DataFrame and 'numeric_cols' is the list of numeric columns
corr_matrix = data[numeric_cols].corr()

# Remove the last two rows and columns
corr_matrix = corr_matrix.iloc[:-2, :-2]

# Create a heatmap with the correlation values
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, cmap='viridis', annot=True, fmt=".2f")
plt.title('Correlation Matrix')
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Additional Analysis

Depending on your requirements, you might want to explore further, such as investigating the relationship between specific features (e.g., customer churn). For example, if the dataset includes an 'Attrition_Flag' column, you can compare numeric features between churned and retained customers.

Below is an example if an 'Attrition_Flag' column exists.

```python
# If 'Attrition_Flag' is in the dataset, compare distributions
if 'Attrition_Flag' in data.columns:
    churned = data[data['Attrition_Flag'] == 'Attrited Customer']
    retained = data[data['Attrition_Flag'] == 'Existing Customer']
    
    plt.figure(figsize=(12, 5))
    plt.hist(churned['Customer_Age'].dropna(), bins=20, alpha=0.5, label='Churned', edgecolor='black')
    plt.hist(retained['Customer_Age'].dropna(), bins=20, alpha=0.5, label='Retained', edgecolor='black')
    plt.title('Customer Age Distribution by Attrition Flag')
    plt.xlabel('Customer Age')
    plt.ylabel('Frequency')
    plt.legend()
    plt.show()
else:
    print("Attrition_Flag column not found in the dataset.")
```
